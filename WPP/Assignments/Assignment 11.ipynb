```python
import pandas as pd
import numpy as np

# Sample DataFrame
data = {
    'John': [True, False, False, True, False, True, False, False, True, False],
    'Judy': [True, False, True, False, True, True, False, False, True, True]
}
df = pd.DataFrame(data)

# Determine party days
party_days = (df['John'] & df['Judy']).values

# Initialize array
days_til_party = [None] * len(df)
next_party = None

# Loop backwards to calculate days until next party
for i in reversed(range(len(df))):
    if party_days[i]:
        next_party = 0
    days_til_party[i] = next_party
    if next_party is not None:
        next_party += 1

# Assign the result as a new column
df['days_til_party'] = days_til_party

print(df)
```

```python
data = {
    'John': [True, False, False, True, False, True, False, False, True, False],
    'Judy': [True, False, True, False, True, True, False, False, True, True]
}
df = pd.DataFrame(data)
party_days = (df['John'] & df['Judy']).values

days_til_party = np.full(len(df), np.inf)

next_party = np.inf

for i in reversed(range(len(df))):

    if party_days[i]:
        next_party = 0
    days_til_party[i] = next_party
    if next_party != np.inf:
        next_party += 1

df['days_til_party'] = days_til_party.astype('int64')

print(df)
```

```python
data = {
    'artist': ['A', 'B', 'A', 'C', 'B', 'A'],
    'venue': ['V1', 'V2', 'V1', 'V1', 'V2', 'V2'],
    'date': pd.to_datetime(['2024-01-01', '2024-01-15', '2024-01-20', '2024-02-05', '2024-02-07', '2024-02-10'])
}
df = pd.DataFrame(data)

df['year_month'] = df['date'].dt.to_period('M')

artists = df['artist'].unique()
venues = df['venue'].unique()
combinations = list(product(artists, venues))

grouped = df.groupby(['year_month', 'artist', 'venue']).size().unstack(level=[1, 2]).fillna(0).astype(int)

for combo in combinations:
    if combo not in grouped.columns:
        grouped[combo] = 0

grouped = grouped.reindex(sorted(grouped.columns), axis=1)
print(grouped)
```
